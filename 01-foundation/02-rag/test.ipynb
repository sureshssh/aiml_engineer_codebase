```python
from openai import OpenAI
```

```python
client = OpenAI()
```

```python
MODEL_PRICE = {
    "gpt-4o-mini":{"input":0.15,"output":0.60},
    "gpt-5-mnano":{"input":0.075,"output":0.30}
}
```

```python
def calculate_price(model_name: str, input_tokens: int, output_tokens: int):
    prices = MODEL_PRICE[model_name.lower()]
    input_cost = (input_tokens/1_000_000) * prices["input"]
    output_cost = (output_tokens/1_000_000)* prices["output"]
    return input_cost + output_cost
```

```python
response = client.responses.create(
    model="gpt-4o-mini",
    input="tell me a joke about Alexey"
)
```

```python
usage = response.usage
cost = calculate_price("gpt-4o-mini",usage.input_tokens, usage.output_tokens)
print(f'cost for this operation {cost}')
```

```python
print(response.model_dump_json(indent=2))
```
